```python
import pandas as pd
```

```python
pm25=pd.read_csv(r"C:\Users\GFD LAB\Documents\Anwesha Chakraborty\output\Delhi_PM25_Average.csv")
blh=pd.read_csv(r"C:\Users\GFD LAB\Documents\Anwesha Chakraborty\output\BLH.csv")
metar=pd.read_csv(r"C:\Users\GFD LAB\Documents\Anwesha Chakraborty\output\metar.csv")
```

```python
pm25["Datetime"] = pd.to_datetime(pm25["Datetime"])
blh["Datetime"] = pd.to_datetime(blh["Datetime"])
metar["Datetime"] = pd.to_datetime(metar["Datetime"])
```

```python
merged = pm25.merge(metar, on="Datetime", how="inner")
merged = merged.merge(blh, on="Datetime", how="inner")
```

```python
merged.isnull().sum()
```

```python
merged = merged.dropna().reset_index(drop=True)
```

```python
merged.to_csv(r"C:\Users\GFD LAB\Documents\Anwesha Chakraborty\output\Final_Dataset.csv", index=False)
```

```python
print("PM25:")
print(pm25["Datetime"].min(), "to", pm25["Datetime"].max(), len(pm25))

print("\nMETAR:")
print(metar["Datetime"].min(), "to", metar["Datetime"].max(), len(metar))

print("\nBLH:")
print(blh["Datetime"].min(), "to", blh["Datetime"].max(), len(blh))
```
