```python
import pandas as pd
```

```python
era5=pd.read_csv(r"C:\Users\GFD LAB\Documents\Anwesha Chakraborty\output\ERA5_Meteorology.csv")
pm25=pd.read_csv(r"C:\Users\GFD LAB\Documents\Anwesha Chakraborty\output\Delhi_PM25_Average.csv")
blh=pd.read_csv(r"C:\Users\GFD LAB\Documents\Anwesha Chakraborty\output\BLH.csv")
metar=pd.read_csv(r"C:\Users\GFD LAB\Documents\Anwesha Chakraborty\output\metar.csv")
```

```python
pm25["Datetime"] = pd.to_datetime(pm25["Datetime"])
era5["Datetime"] = pd.to_datetime(era5["Datetime"])
blh["Datetime"] = pd.to_datetime(blh["Datetime"])
metar["Datetime"] = pd.to_datetime(metar["Datetime"])
```

```python
metar = metar[["Datetime", "Visibility"]]
```

```python
df = pm25.merge(era5, on="Datetime", how="inner")
```

```python
df = df.merge(blh,on="Datetime",how="inner")
```

```python
df = df.merge(metar, on="Datetime",how="inner")
```

```python
df = df[
    [
        "Datetime",
        "PM2.5_Mean",
        "Visibility",
        "Temperature",
        "RH",
        "WindSpeed",
        "BLH"
    ]
]
```

```python
df = df.dropna().reset_index(drop=True)
```

```python
df.to_csv(r"C:\Users\GFD LAB\Documents\Anwesha Chakraborty\output\Final_Dataset_ERA5.csv",index=False)
```
