```python
import pandas as pd
```

```python
metar=pd.read_csv( r"C:\Users\GFD LAB\Documents\Anwesha Chakraborty\data\visibility, wind speed, rh, temp\VIDP (1).csv", skiprows=5)
```

```python
metar["valid"] = pd.to_datetime(
    metar["valid"],
    dayfirst=False
)
```

```python
metar = metar[metar["valid"].dt.minute == 0]
```

```python
metar = metar[["valid", "vsby", "relh", "tmpc", "sped"]]
```

```python
metar.rename(columns={
    "valid": "Datetime",
    "vsby": "Visibility",
    "relh": "RH",
    "tmpc": "Temperature",
    "sped": "WindSpeed"
}, inplace=True)
```

```python
metar["Visibility"] = metar["Visibility"] * 1609.344
```

```python
metar["WindSpeed"] = metar["WindSpeed"] * 0.44704
```

```python
metar = metar.sort_values("Datetime").reset_index(drop=True)
```

```python
metar.to_csv(r"C:\Users\GFD LAB\Documents\Anwesha Chakraborty\output\metar.csv",index=False)
```
