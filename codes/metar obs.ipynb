```python
metar = pd.read_csv(r"C:\Users\GFD LAB\Documents\Anwesha Chakraborty\data\visibility, wind speed, rh, temp\VIDP (1).csv",skiprows=5)
```

```python
print(metar.head(10))
```

```python
metar["valid"] = pd.to_datetime(metar["valid"])
```

```python
metar = metar[metar["valid"].dt.minute == 0]
```

```python
metar = metar[["valid", "vsby", "relh", "tmpc", "sped"]]
```

```python
print(metar.head(10))
```

```python
print(metar.dtypes)
```

```python
metar["vsby"] = metar["vsby"] * 1609.344   #visibility from miles to metres
```

```python
metar["sped"] = metar["sped"] * 0.44704  
```

```python
metar.rename(columns={
    "valid": "Datetime",
    "vsby": "Visibility",
    "relh": "RH",
    "tmpc": "Temperature",
    "sped": "WindSpeed"
}, inplace=True)
```

```python
metar = metar[
    ["Datetime", "Visibility", "RH", "Temperature", "WindSpeed"]
]
```

```python
metar = metar.sort_values("Datetime").reset_index(drop=True)
```
