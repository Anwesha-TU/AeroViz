```python
df = pd.read_csv(
    r"C:\Users\GFD LAB\Documents\Anwesha Chakraborty\output\Final_Dataset.csv"
)
df["Datetime"] = pd.to_datetime(df["Datetime"])
monthly = df.set_index("Datetime").resample("ME").mean(numeric_only=True)
```

```python
plt.figure(figsize=(12,5))
plt.plot(monthly.index, monthly["Temperature"], linewidth=2)
plt.title("Monthly Mean Temperature")
plt.xlabel("Year")
plt.ylabel("Temperature (°C)")
plt.grid(True)
plt.show()

plt.figure(figsize=(12,5))
plt.plot(monthly.index, monthly["RH"], linewidth=2)
plt.title("Monthly Mean Relative Humidity")
plt.xlabel("Year")
plt.ylabel("RH (%)")
plt.grid(True)
plt.show()

plt.figure(figsize=(12,5))
plt.plot(monthly.index, monthly["WindSpeed"], linewidth=2)
plt.title("Monthly Mean Wind Speed")
plt.xlabel("Year")
plt.ylabel("Wind Speed (m/s)")
plt.grid(True)
plt.show()

plt.figure(figsize=(12,5))
plt.plot(monthly.index, monthly["Visibility"], linewidth=2)
plt.title("Monthly Mean Visibility")
plt.xlabel("Year")
plt.ylabel("Visibility (m)")
plt.grid(True)
plt.show()
```

```python
df["Datetime"] = pd.to_datetime(df["Datetime"])
low_visibility = df[df["Visibility"] < 1000].copy()
low_visibility["Date"] = low_visibility["Datetime"].dt.date
low_visibility["Month"] = low_visibility["Datetime"].dt.to_period("M")
monthly_low_visibility = (
    low_visibility.groupby("Month")["Date"]
    .nunique()
    .reset_index(name="Low_Visibility_Days")
)
print(monthly_low_visibility)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 5))

plt.bar(
    monthly_low_visibility["Month"].astype(str),
    monthly_low_visibility["Low_Visibility_Days"]
)

plt.title("Monthly Number of Days with Visibility < 1000 m")
plt.xlabel("Month")
plt.ylabel("Number of Days")

plt.xticks(rotation=90)
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.savefig(
    r"C:\Users\GFD LAB\Desktop\monthly_low_visibility_days.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
monthly_low_visibility["Month"] = pd.to_datetime(
    monthly_low_visibility["Month"].astype(str)
)
monthly_low_visibility["Month_Number"] = (
    monthly_low_visibility["Month"].dt.month
)
monthly_average = (
    monthly_low_visibility
    .groupby("Month_Number")["Low_Visibility_Days"]
    .mean()
    .reset_index()
)
monthly_average["Month"] = pd.to_datetime(
    monthly_average["Month_Number"],
    format="%m"
).dt.strftime("%B")
print(monthly_average)
```

```python
plt.figure(figsize=(12,5))

plt.bar(
    monthly_average["Month"],
    monthly_average["Low_Visibility_Days"]
)

plt.title("Average Number of Days with Visibility < 1000 m")
plt.xlabel("Month")
plt.ylabel("Average Number of Days")

plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.savefig(
    r"C:\Users\GFD LAB\Desktop\monthly_low_visibility_days2.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```
