```python
import matplotlib.pyplot as plt
labels = {
    "PM2.5_Mean": "PM2.5 Mean (µg/m³)",
    "Visibility": "Visibility (m)",      # Change to (m) if your dataset is in meters
    "RH": "Relative Humidity (%)",
    "Temperature": "Temperature (°C)",
    "WindSpeed": "Wind Speed (m/s)",      # Change if your dataset uses km/h
    "BLH": "Boundary Layer Height (m)"
}

columns = [
    "PM2.5_Mean",
    "Visibility",
    "RH",
    "Temperature",
    "WindSpeed",
    "BLH"
]

for col in columns:
    plt.figure(figsize=(6,4))
    plt.hist(df[col], bins=40, edgecolor="black")
    plt.title(f"Histogram of {labels[col]}")
    plt.xlabel(labels[col], fontsize=11)
    plt.ylabel("Frequency", fontsize=11)
    plt.grid(axis='y', linestyle='--', alpha=0.5)
    plt.tight_layout()
    plt.show()
```

```python
#box plots to detect outliers
import matplotlib.pyplot as plt
labels = {
    "PM2.5_Mean": "PM2.5 Mean (µg/m³)",
    "Visibility": "Visibility (m)",
    "RH": "Relative Humidity (%)",
    "Temperature": "Temperature (°C)",
    "WindSpeed": "Wind Speed (m/s)",
    "BLH": "Boundary Layer Height (m)"
}

columns = [
    "PM2.5_Mean",
    "Visibility",
    "RH",
    "Temperature",
    "WindSpeed",
    "BLH"
]

for col in columns:
    plt.figure(figsize=(6, 4))
    plt.boxplot(df[col], vert=True)
    plt.title(f"Boxplot of {labels[col]}")
    plt.ylabel(labels[col])
    plt.grid(axis='y', linestyle='--', alpha=0.5)
    plt.tight_layout()
    plt.show()
```

```python
corr_matrix = df.corr(numeric_only=True)
corr_matrix
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(8,6))
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    square=True
)
plt.title("Correlation Matrix of Air Quality and Meteorological Variables")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
labels = {
    "PM2.5_Mean": "PM2.5 Mean (µg/m³)",
    "RH": "Relative Humidity (%)",
    "Temperature": "Temperature (°C)",
    "WindSpeed": "Wind Speed (m/s)",
    "BLH": "Boundary Layer Height (m)"
}
features = [
    "PM2.5_Mean",
    "RH",
    "Temperature",
    "WindSpeed",
    "BLH"
]
for feature in features:
    plt.figure(figsize=(6,4))
    plt.scatter(df[feature], df["Visibility"],
                alpha=0.5,
                s=20)
    plt.xlabel(labels[feature])
    plt.ylabel("Visibility (m)")   # or km depending on your dataset
    plt.title(f"Visibility vs {labels[feature]}")
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.tight_layout()
    plt.show()
```
