```python
import xarray as xr
file1=xr.open_dataset(r"C:\Users\GFD LAB\Documents\Anwesha Chakraborty\data\blh\aa4e6f661a81acabf7c9dc3789bb20d3.nc")
file2=xr.open_dataset(r"C:\Users\GFD LAB\Documents\Anwesha Chakraborty\data\blh\b9c0dd3c8879267b43868f582c5b0e79.nc")
```

```python
output_csv = r"C:\Users\GFD LAB\Documents\Anwesha Chakraborty\output\blh.csv"
```

```python
ds = xr.concat([file1, file2], dim="time")
```

```python
blh = ds["blh"].mean(dim=["latitude", "longitude"])
```

```python
df = blh.to_dataframe().reset_index()
```

```python
df = df[["valid_time","blh"]]
```

```python
df.rename(columns={
    "valid_time":"Datetime",
    "blh":"BLH"
}, inplace=True)
```

```python
import pandas as pd
```

```python
df.to_csv(output_csv, index=False)
```
