```python
input_folder = r"C:\Users\GFD LAB\Documents\Anwesha Chakraborty\data\pm 2.5"

merged_station_folder = r"C:\Users\GFD LAB\Documents\Anwesha Chakraborty\merged stations"
output_folder = r"C:\Users\GFD LAB\Documents\Anwesha Chakraborty\output"
```

```python
station_data = defaultdict(list)
report = []

excel_files = [f for f in os.listdir(input_folder) if f.endswith(".xlsx")]

print(f"Found {len(excel_files)} Excel files.\n")
```

```python
for file in excel_files:

    filepath = os.path.join(input_folder, file)

    try:
        # Read metadata
        metadata = pd.read_excel(filepath, header=None, nrows=10)
        station = metadata.iloc[5, 2]
        if pd.isna(station):
            station = "Unknown Station"
        station = str(station).strip()
        # Make filename-safe
        station_filename = re.sub(r'[\\/*?:"<>|]', "_", station)
        # Read PM2.5 data
        df = pd.read_excel(filepath, skiprows=10)
        df = df[["From Date", "PM2.5"]].copy()
        df.rename(columns={
                "From Date": "Datetime",
                "PM2.5": "PM2.5"
            },
            inplace=True
        )

        df["Datetime"] = pd.to_datetime(df["Datetime"],dayfirst=True,errors="coerce")
        df["PM2.5"] = pd.to_numeric(df["PM2.5"],errors="coerce")
        before = len(df)     
        df.dropna(subset=["Datetime"], inplace=True)
        duplicates = df.duplicated(subset="Datetime").sum()
        missing = df["PM2.5"].isna().sum()
        station_data[station].append(df)
        report.append({
            "Station": station,
            "File": file,
            "Rows": before,
            "Duplicate Hours": duplicates,
            "Missing PM2.5": missing
        })
        print(f"Processed : {file}")
    except Exception as e:
        print(f"Error in {file}")
        print(e)
```

```python
master_df = None

print("\nMerging station files...\n")

for station, dfs in station_data.items():

    station_filename = re.sub(r'[\\/*?:"<>|]', "_", station)

    station_df = pd.concat(
        dfs,
        ignore_index=True
    )

    station_df.sort_values(
        "Datetime",
        inplace=True
    )

    station_df.drop_duplicates(
        subset="Datetime",
        keep="first",
        inplace=True
    )

    station_df.to_csv(
        os.path.join(
            merged_station_folder,
            f"{station_filename}.csv"
        ),
        index=False
    )

    temp = station_df.rename(
        columns={
            "PM2.5": station
        }
    )

    if master_df is None:
        master_df = temp
    else:
        master_df = pd.merge(
            master_df,
            temp,
            on="Datetime",
            how="outer"
        )
```

```python
#sort master data 
master_df.sort_values(
    "Datetime",
    inplace=True
)

master_df.reset_index(
    drop=True,
    inplace=True
)
```

```python
#save master dataset
master_df.to_csv(
    os.path.join(
        output_folder,
        "PM25_All_Stations.csv"
    ),
    index=False
)
```

```python
avg_df = master_df.copy()
station_columns = avg_df.columns.drop("Datetime")
avg_df["PM2.5_Mean"] = avg_df[
    station_columns
].mean(
    axis=1,
    skipna=True
)
avg_df = avg_df[
    ["Datetime", "PM2.5_Mean"]
]
average_path = os.path.join(
    output_folder,
    "Delhi_PM25_Average.csv"
)
avg_df.to_csv(
    average_path,
    index=False
)
```

```python
report_df = pd.DataFrame(report)

summary = report_df.groupby("Station").agg({
    "File": "count",
    "Rows": "sum",
    "Duplicate Hours": "sum",
    "Missing PM2.5": "sum"
}).reset_index()
```

```python
summary.rename(
    columns={
        "File": "Files Processed"
    },
    inplace=True
)

report_path = os.path.join(
    output_folder,
    "Processing_Report.csv"
)

summary.to_csv(
    report_path,
    index=False
)
```
