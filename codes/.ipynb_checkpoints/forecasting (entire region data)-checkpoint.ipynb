```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.svm import SVR
```

```python
from sklearn.metrics import (mean_absolute_error, mean_squared_error, r2_score)
```

```python
df=pd.read_csv(r"C:\Users\acann\OneDrive\Desktop\mtech major project\Anwesha Chakraborty\output\Final_Dataset_ERA5.csv")
```

```python
df["Datetime"] = pd.to_datetime(df["Datetime"])
```

```python
features = [
    "PM2.5_Mean",
    "RH",
    "Temperature",
    "WindSpeed",
    "BLH"
]

target = "Visibility"
```

```python
for h in range(1, 13):
    df[f"Visibility_{h}h"] = df[target].shift(-h)
    df[f"Target_Time_{h}h"] = df["Datetime"].shift(-h)
```

```python
target_columns = [
    f"Visibility_{h}h"
    for h in range(1, 13)
]

df_forecast = df.dropna(
    subset=target_columns
).copy()

print("Forecasting dataset shape:", df_forecast.shape)
```

```python
fog_months = [1, 2, 3, 11, 12]

case1_train_mask = (
    (df_forecast["Datetime"].dt.year.isin([2024, 2025])) &
    (df_forecast["Datetime"].dt.month.isin(fog_months)))

print(
    "Case 1 training samples:",
    case1_train_mask.sum()
)
```

```python
case2_train_mask = (
    df_forecast["Datetime"].dt.year.isin([2024, 2025]))
print(
    "Case 2 training samples:",
    case2_train_mask.sum()
)
```

```python
def get_test_mask(df, lead_time):

    target_time_col = f"Target_Time_{lead_time}h"

    test_mask = (
        (df[target_time_col] >= "2026-01-01") &
        (df[target_time_col] < "2026-04-01")
    )

    return test_mask
```

```python
def get_models():

    models = {

        "Linear Regression": {
            "model": LinearRegression(),
            "scale": True
        },

        "Ridge Regression": {
            "model": Ridge(alpha=1),
            "scale": True
        },

        "Random Forest": {
            "model": RandomForestRegressor(
                n_estimators=300,
                random_state=42,
                n_jobs=-1
            ),
            "scale": False
        },

        "Gradient Boosting": {
            "model": GradientBoostingRegressor(
                n_estimators=300,
                random_state=42
            ),
            "scale": False
        },

        "SVR": {
            "model": SVR(
                kernel="rbf",
                C=100,
                epsilon=0.1
            ),
            "scale": True
        }
    }

    return models
```

```python
def run_forecast(
    df,
    model,
    scale_features,
    features,
    lead_time,
    train_mask,
    test_mask
):

    target_column = f"Visibility_{lead_time}h"

    X_train = df.loc[
        train_mask,
        features
    ].copy()

    y_train = df.loc[
        train_mask,
        target_column
    ].copy()

    X_test = df.loc[
        test_mask,
        features
    ].copy()

    y_test = df.loc[
        test_mask,
        target_column
    ].copy()


    train_data = pd.concat(
        [X_train, y_train],
        axis=1
    ).dropna()

    test_data = pd.concat(
        [X_test, y_test],
        axis=1
    ).dropna()


    X_train = train_data[features]
    y_train = train_data[target_column]

    X_test = test_data[features]
    y_test = test_data[target_column]


    if scale_features:

        scaler = StandardScaler()

        X_train_model = scaler.fit_transform(
            X_train
        )

        X_test_model = scaler.transform(
            X_test
        )

    else:

        X_train_model = X_train
        X_test_model = X_test

    model.fit(
        X_train_model,
        y_train
    )

    y_pred = model.predict(
        X_test_model
    )

    mae = mean_absolute_error(
        y_test,
        y_pred
    )

    mse = mean_squared_error(
        y_test,
        y_pred
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        y_test,
        y_pred
    )
    return {
        "MAE": mae, "MSE": mse, "RMSE": rmse, "R2": r2, "Actual": y_test, "Predicted": y_pred
    }
```

```python
lead_times = list(range(1, 13))
print("Lead times:", lead_times)
```

```python
training_cases = {
    "Case 1 - Fog Season": case1_train_mask,
    "Case 2 - All Months": case2_train_mask
}
```

```python
models = get_models()
all_forecast_results = []
all_forecast_predictions = {}
```

```python
for case_name, train_mask in training_cases.items():

    for lead_time in lead_times:

        # Test mask for this lead time
        test_mask = get_test_mask(
            df_forecast,
            lead_time
        )
        for model_name, model_info in models.items():
            print(
                f"Running: "
                f"{model_name} | "
                f"{case_name} | "
                f"{lead_time}-hour forecast"
            )
            result = run_forecast(
                df=df_forecast,
                model=model_info["model"],
                scale_features=model_info["scale"],
                features=features,
                lead_time=lead_time,
                train_mask=train_mask,
                test_mask=test_mask
            )
            # Store numerical results
            all_forecast_results.append({
                "Model": model_name,
                "Training Case": case_name,
                "Lead Time (hours)": lead_time,
                "MAE": result["MAE"],
                "MSE": result["MSE"],
                "RMSE": result["RMSE"],
                "R2": result["R2"]
            })
            # Store predictions
            prediction_key = (
                f"{model_name}_"
                f"{case_name}_"
                f"{lead_time}h"
            )
            all_forecast_predictions[
                prediction_key
            ] = {
                "Actual": result["Actual"],
                "Predicted": result["Predicted"]
            }
```

```python
from sklearn.base import clone
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
base_features = [
    "PM2.5_Mean",
    "RH",
    "Temperature",
    "WindSpeed",
    "BLH"
]

target = "Visibility"

lead_times = list(range(1, 13))

fog_months = [11, 12, 1, 2, 3]

print("Base features:")
print(base_features)

print("\nForecast horizons:")
print(lead_times)
```

```python
def create_window_features(df, features, window):

    data = df.copy()

    for feature in features:

        # Current hour
        data[f"{feature}_t"] = data[feature]

        # Previous hours
        for lag in range(1, window):
            data[f"{feature}_lag{lag}"] = data[feature].shift(lag)

    return data
```

```python
def create_targets(df, target):

    data = df.copy()

    for h in lead_times:

        data[f"{target}_{h}h"] = (
            data[target].shift(-h)
        )

        data[f"Target_Time_{h}h"] = (
            data["Datetime"].shift(-h)
        )

    return data
```

```python
df_12h = create_window_features(
    df,
    base_features,
    window=12
)

df_12h = create_targets(
    df_12h,
    target
)

print("12-hour window shape:", df_12h.shape)
```

```python
df_24h = create_window_features(
    df,
    base_features,
    window=24
)

df_24h = create_targets(
    df_24h,
    target
)

print("24-hour window shape:", df_24h.shape)
```

```python
def get_window_features(features, window):

    feature_columns = []

    for feature in features:

        feature_columns.append(
            f"{feature}_t"
        )

        for lag in range(1, window):

            feature_columns.append(
                f"{feature}_lag{lag}"
            )

    return feature_columns


features_12h = get_window_features(
    base_features,
    12
)

features_24h = get_window_features(
    base_features,
    24
)

print(
    "12-hour input features:",
    len(features_12h)
)

print(
    "24-hour input features:",
    len(features_24h)
)
```

```python
all_months_train_mask = (
    df_12h["Datetime"].dt.year.isin([2024, 2025])
)

winter_train_mask = (
    df_12h["Datetime"].dt.year.isin([2024, 2025])
    &
    df_12h["Datetime"].dt.month.isin(fog_months)
)

print(
    "All-month training samples:",
    all_months_train_mask.sum()
)

print(
    "Fog season training samples:",
    winter_train_mask.sum()
)
```

```python
def get_test_mask(df, lead_time):

    target_time_col = f"Target_Time_{lead_time}h"

    return (
        (df[target_time_col] >= "2026-01-01")
        &
        (df[target_time_col] < "2026-04-01")
    )
```

```python
def get_models():

    return {

        "Linear Regression": {
            "model": LinearRegression(),
            "scale": True
        },

        "Ridge Regression": {
            "model": Ridge(alpha=1),
            "scale": True
        },

        "Random Forest": {
            "model": RandomForestRegressor(
                n_estimators=300,
                random_state=42,
                n_jobs=-1
            ),
            "scale": False
        },

        "Gradient Boosting": {
            "model": GradientBoostingRegressor(
                n_estimators=300,
                random_state=42
            ),
            "scale": False
        },

        "SVR": {
            "model": SVR(
                kernel="rbf",
                C=100,
                epsilon=0.1
            ),
            "scale": True
        }
    }


models = get_models()
```

```python
def run_window_forecast(
    data,
    features,
    lead_time,
    train_mask,
    model,
    scale_features
):

    target_column = f"Visibility_{lead_time}h"

    test_mask = get_test_mask(
        data,
        lead_time
    )

    train_data = data.loc[
        train_mask,
        features + [target_column]
    ].dropna()

    X_train = train_data[features]
    y_train = train_data[target_column]

    test_data = data.loc[
        test_mask,
        features + [target_column]
    ].dropna()

    X_test = test_data[features]
    y_test = test_data[target_column]
    if scale_features:

        scaler = StandardScaler()

        X_train_model = scaler.fit_transform(
            X_train
        )

        X_test_model = scaler.transform(
            X_test
        )

    else:

        X_train_model = X_train
        X_test_model = X_test

    model = clone(model)

    model.fit(
        X_train_model,
        y_train
    )

    y_pred = model.predict(
        X_test_model
    )

    mae = mean_absolute_error(
        y_test,
        y_pred
    )

    mse = mean_squared_error(
        y_test,
        y_pred
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        y_test,
        y_pred
    )

    return {
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2": r2,
        "Actual": y_test,
        "Predicted": y_pred
    }
```

```python
results_12h = []
predictions_12h = {}

training_cases_12h = {
    "12h - All Months": all_months_train_mask,
    "12h - Fog season Months": winter_train_mask
}

for case_name, train_mask in training_cases_12h.items():

    for lead_time in lead_times:

        for model_name, model_info in models.items():

            print(
                f"Running: {case_name} | "
                f"{model_name} | "
                f"{lead_time}h"
            )

            result = run_window_forecast(
                data=df_12h,
                features=features_12h,
                lead_time=lead_time,
                train_mask=train_mask,
                model=model_info["model"],
                scale_features=model_info["scale"]
            )

            results_12h.append({
                "Training Case": case_name,
                "Model": model_name,
                "Lead Time (hours)": lead_time,
                "MAE": result["MAE"],
                "MSE": result["MSE"],
                "RMSE": result["RMSE"],
                "R2": result["R2"]
            })

            predictions_12h[
                f"{case_name}_{model_name}_{lead_time}h"
            ] = {
                "Actual": result["Actual"],
                "Predicted": result["Predicted"]
            }


results_12h_df = pd.DataFrame(
    results_12h
).round(3)

results_12h_df
```

```python
results_24h = []
predictions_24h = {}

training_cases_24h = {
    "24h - All Months": all_months_train_mask,
    "24h - Fog Season Months": winter_train_mask
}

for case_name, train_mask in training_cases_24h.items():

    for lead_time in lead_times:

        for model_name, model_info in models.items():

            print(
                f"Running: {case_name} | "
                f"{model_name} | "
                f"{lead_time}h"
            )

            result = run_window_forecast(
                data=df_24h,
                features=features_24h,
                lead_time=lead_time,
                train_mask=train_mask,
                model=model_info["model"],
                scale_features=model_info["scale"]
            )

            results_24h.append({
                "Training Case": case_name,
                "Model": model_name,
                "Lead Time (hours)": lead_time,
                "MAE": result["MAE"],
                "MSE": result["MSE"],
                "RMSE": result["RMSE"],
                "R2": result["R2"]
            })

            predictions_24h[
                f"{case_name}_{model_name}_{lead_time}h"
            ] = {
                "Actual": result["Actual"],
                "Predicted": result["Predicted"]
            }


results_24h_df = pd.DataFrame(
    results_24h
).round(3)

results_24h_df
```

```python
window_results_df = pd.concat(
    [
        results_12h_df,
        results_24h_df
    ],
    ignore_index=True
)

window_results_df = window_results_df.round(3)

window_results_df
```

```python
best_window_models = (
    window_results_df
    .loc[
        window_results_df
        .groupby(
            [
                "Training Case",
                "Lead Time (hours)"
            ]
        )["RMSE"]
        .idxmin()
    ]
    .sort_values(
        [
            "Training Case",
            "Lead Time (hours)"
        ]
    )
    .reset_index(drop=True)
)

best_window_models.round(3)
```

```python
LOW_VIS_THRESHOLD = 1000 
```

```python
from sklearn.metrics import confusion_matrix, recall_score, precision_score
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.metrics import (
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score
)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


def calculate_event_metrics(actual, predicted):

    actual_event = (
        np.asarray(actual) < LOW_VIS_THRESHOLD
    ).astype(int)

    predicted_event = (
        np.asarray(predicted) < LOW_VIS_THRESHOLD
    ).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        actual_event,
        predicted_event,
        labels=[0, 1]
    ).ravel()

    precision = precision_score(
        actual_event,
        predicted_event,
        zero_division=0
    )

    recall = recall_score(
        actual_event,
        predicted_event,
        zero_division=0
    )

    f1 = f1_score(
        actual_event,
        predicted_event,
        zero_division=0
    )

    return {
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp,
        "Precision": precision,
        "Recall": recall,
        "F1 Score": f1
    }
```

```python
selected_models = [
    "Gradient Boosting",
    "SVR"
]

case_0h = "Case 1 - Fog Season"
case_12h = "12h - Fog Season"
case_24h = "24h - Fog Season"

lead = 1

event_results = []

prediction_sets = {
    "0-hour": (
        all_forecast_predictions,
        lambda model: (
            f"{model}_{case_0h}_{lead}h"
        )
    ),

    "12-hour": (
        predictions_12h,
        lambda model: (
            f"{case_12h}_{model}_{lead}h"
        )
    ),

    "24-hour": (
        predictions_24h,
        lambda model: (
            f"{case_24h}_{model}_{lead}h"
        )
    )
}
```

```python
for window_name, (
    prediction_dict,
    key_function
) in prediction_sets.items():

    for model in selected_models:

        key = key_function(model)

        if key not in prediction_dict:

            print(
                f"Missing prediction: {key}"
            )

            continue

        actual = prediction_dict[key]["Actual"]
        predicted = prediction_dict[key]["Predicted"]

        metrics = calculate_event_metrics(
            actual,
            predicted
        )

        event_results.append({
            "Window": window_name,
            "Model": model,
            "Lead Time (hours)": lead,
            **metrics
        })


event_results_df = pd.DataFrame(
    event_results
)

event_results_df[
    [
        "Window",
        "Model",
        "Lead Time (hours)",
        "TP",
        "FP",
        "FN",
        "TN",
        "Precision",
        "Recall",
        "F1 Score"
    ]
].round(3)
```

```python
def plot_cm(
    actual,
    predicted,
    title
):

    actual_event = (
        np.asarray(actual) < LOW_VIS_THRESHOLD
    ).astype(int)

    predicted_event = (
        np.asarray(predicted) < LOW_VIS_THRESHOLD
    ).astype(int)

    cm = confusion_matrix(
        actual_event,
        predicted_event,
        labels=[0, 1]
    )

    plt.figure(figsize=(6, 5))

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        xticklabels=[
            "Normal Visibility",
            "Low Visibility"
        ],
        yticklabels=[
            "Normal Visibility",
            "Low Visibility"
        ]
    )

    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.title(title)

    plt.tight_layout()
    plt.show()
```

```python
from sklearn.metrics import (
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score
)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

LOW_VIS_THRESHOLD = 1000  # metres

selected_models = [
    "Gradient Boosting",
    "SVR"
]
```

```python
prediction_sources = {
    
    "0-hour": {
        "predictions": all_forecast_predictions,
        "case": "Case 1 - Fog Season",
        "key_order": "model_first"
    },

    "12-hour": {
        "predictions": predictions_12h,
        "case": "12h - Fog Season",
        "key_order": "case_first"
    },

    "24-hour": {
        "predictions": predictions_24h,
        "case": "24h - Fog Season",
        "key_order": "case_first"
    }
}
```

```python
event_results = []

for window_name, info in prediction_sources.items():

    prediction_dict = info["predictions"]
    case = info["case"]

    for model in selected_models:

        for lead in range(1, 13):

            # -----------------------------
            # Construct prediction key
            # -----------------------------

            if info["key_order"] == "model_first":

                key = (
                    f"{model}_"
                    f"{case}_"
                    f"{lead}h"
                )

            else:

                key = (
                    f"{case}_"
                    f"{model}_"
                    f"{lead}h"
                )

            # -----------------------------
            # Check key
            # -----------------------------

            if key not in prediction_dict:

                print("Missing:", key)
                continue

            # -----------------------------
            # Get predictions
            # -----------------------------

            actual = np.asarray(
                prediction_dict[key]["Actual"]
            )

            predicted = np.asarray(
                prediction_dict[key]["Predicted"]
            )

            # -----------------------------
            # Convert visibility to event
            # -----------------------------

            actual_event = (
                actual < LOW_VIS_THRESHOLD
            ).astype(int)

            predicted_event = (
                predicted < LOW_VIS_THRESHOLD
            ).astype(int)

            # -----------------------------
            # Confusion matrix
            # -----------------------------

            tn, fp, fn, tp = confusion_matrix(
                actual_event,
                predicted_event,
                labels=[0, 1]
            ).ravel()

            # -----------------------------
            # Metrics
            # -----------------------------

            precision = precision_score(
                actual_event,
                predicted_event,
                zero_division=0
            )

            recall = recall_score(
                actual_event,
                predicted_event,
                zero_division=0
            )

            f1 = f1_score(
                actual_event,
                predicted_event,
                zero_division=0
            )

            # -----------------------------
            # Store
            # -----------------------------

            event_results.append({

                "Window": window_name,

                "Model": model,

                "Lead Time (hours)": lead,

                "TP": tp,

                "FP": fp,

                "FN": fn,

                "TN": tn,

                "Precision": precision,

                "Recall": recall,

                "F1 Score": f1
            })


event_results_df = pd.DataFrame(
    event_results
)

event_results_df = event_results_df.round(4)

event_results_df
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score
)

LOW_VIS_THRESHOLD = 1000  # metres

selected_models = ["Gradient Boosting", "SVR"]

# Check that the model prediction dictionaries exist.
required_variables = [
    "all_forecast_predictions",
    "predictions_12h",
    "predictions_24h"
]

missing_variables = [
    name for name in required_variables
    if name not in globals()
]

if missing_variables:
    raise NameError(
        f"Missing prediction variables: {missing_variables}. "
        "Run the model-training/prediction cells that create them first."
    )

print("Original prediction entries:", len(all_forecast_predictions))
print("12-hour prediction entries:", len(predictions_12h))
print("24-hour prediction entries:", len(predictions_24h))
print("Low-visibility threshold:", LOW_VIS_THRESHOLD, "m")
```

```python

prediction_sources = {
    "0-hour - Fog Season": {
        "predictions": all_forecast_predictions,
        "case": "Case 1 - Fog Season",
        "key_order": "model_first"
    },
    "0-hour - All Months": {
        "predictions": all_forecast_predictions,
        "case": "Case 2 - All Months",
        "key_order": "model_first"
    },
    "12-hour - Fog Season": {
        "predictions": predictions_12h,
        "case": "12h - Fog season Months",
        "key_order": "case_first"
    },
    "12-hour - All Months": {
        "predictions": predictions_12h,
        "case": "12h - All Months",
        "key_order": "case_first"
    },
    "24-hour - Fog Season": {
        "predictions": predictions_24h,
        "case": "24h - Fog Season Months",
        "key_order": "case_first"
    },
    "24-hour - All Months": {
        "predictions": predictions_24h,
        "case": "24h - All Months",
        "key_order": "case_first"
    }
}

print("Evaluation groups:")
for window_name in prediction_sources:
    print("-", window_name)
```

```python

def calculate_event_metrics(actual, predicted):
    actual = np.asarray(actual, dtype=float).ravel()
    predicted = np.asarray(predicted, dtype=float).ravel()

    # Remove pairs where either value is missing or non-finite.
    valid = np.isfinite(actual) & np.isfinite(predicted)
    actual = actual[valid]
    predicted = predicted[valid]

    if len(actual) == 0:
        return None

    actual_event = (actual < LOW_VIS_THRESHOLD).astype(int)
    predicted_event = (predicted < LOW_VIS_THRESHOLD).astype(int)

    tn, fp, fn, tp = confusion_matrix(
        actual_event,
        predicted_event,
        labels=[0, 1]
    ).ravel()

    return {
        "Samples": len(actual),
        "TP": int(tp),
        "FP": int(fp),
        "FN": int(fn),
        "TN": int(tn),
        "Precision": precision_score(
            actual_event, predicted_event, zero_division=0
        ),
        "Recall": recall_score(
            actual_event, predicted_event, zero_division=0
        ),
        "F1 Score": f1_score(
            actual_event, predicted_event, zero_division=0
        )
    }


event_rows = []
missing_prediction_keys = []

for window_name, info in prediction_sources.items():
    prediction_dict = info["predictions"]
    case = info["case"]

    for model in selected_models:
        for lead in range(1, 13):

            if info["key_order"] == "model_first":
                key = f"{model}_{case}_{lead}h"
            else:
                key = f"{case}_{model}_{lead}h"

            if key not in prediction_dict:
                missing_prediction_keys.append((window_name, key))
                continue

            prediction = prediction_dict[key]

            if "Actual" not in prediction or "Predicted" not in prediction:
                print(f"Skipping malformed prediction entry: {key}")
                continue

            metrics = calculate_event_metrics(
                prediction["Actual"],
                prediction["Predicted"]
            )

            if metrics is None:
                print(f"Skipping entry with no valid samples: {key}")
                continue

            event_rows.append({
                "Window": window_name,
                "Model": model,
                "Lead Time (hours)": lead,
                **metrics
            })

event_results_df = pd.DataFrame(event_rows)

if event_results_df.empty:
    raise ValueError(
        "No event metrics were calculated. Check the prediction keys "
        "printed in the diagnostics below."
    )

event_results_df = event_results_df.sort_values(
    ["Window", "Model", "Lead Time (hours)"]
).reset_index(drop=True)

print("Metrics rows generated:", len(event_results_df))
print("Expected rows if all predictions exist:", 6 * 2 * 12)
print("Missing prediction keys:", len(missing_prediction_keys))

for window_name, key in missing_prediction_keys:
    print(f"Missing [{window_name}]: {key}")

display(event_results_df.round(4))
```

```python

def plot_all_confusion_matrices(window_name, model):
    info = prediction_sources[window_name]
    prediction_dict = info["predictions"]
    case = info["case"]

    fig, axes = plt.subplots(3, 4, figsize=(16, 11))
    axes = axes.flatten()
    plotted = 0

    for i, lead in enumerate(range(1, 13)):
        if info["key_order"] == "model_first":
            key = f"{model}_{case}_{lead}h"
        else:
            key = f"{case}_{model}_{lead}h"

        ax = axes[i]

        if key not in prediction_dict:
            ax.set_title(f"{lead}h — prediction missing")
            ax.axis("off")
            continue

        prediction = prediction_dict[key]
        actual = np.asarray(prediction["Actual"], dtype=float).ravel()
        predicted = np.asarray(prediction["Predicted"], dtype=float).ravel()

        valid = np.isfinite(actual) & np.isfinite(predicted)
        actual = actual[valid]
        predicted = predicted[valid]

        if len(actual) == 0:
            ax.set_title(f"{lead}h — no valid samples")
            ax.axis("off")
            continue

        actual_event = (actual < LOW_VIS_THRESHOLD).astype(int)
        predicted_event = (predicted < LOW_VIS_THRESHOLD).astype(int)

        cm = confusion_matrix(
            actual_event,
            predicted_event,
            labels=[0, 1]
        )

        sns.heatmap(
            cm,
            annot=True,
            fmt="d",
            cmap="Blues",
            cbar=False,
            ax=ax,
            xticklabels=["Normal", "Low"],
            yticklabels=["Normal", "Low"]
        )

        ax.set_title(f"{lead}-Hour Forecast")
        ax.set_xlabel("Predicted class")
        ax.set_ylabel("Actual class")
        plotted += 1

    fig.suptitle(
        f"{model} | {window_name}\n"
        f"Low visibility: < {LOW_VIS_THRESHOLD} m",
        fontsize=15,
        y=1.02
    )

    plt.tight_layout()
    plt.show()

    print(f"{window_name} | {model}: plotted {plotted}/12 matrices")


for window_name in prediction_sources:
    for model in selected_models:
        plot_all_confusion_matrices(window_name, model)
```

```python

def plot_event_metric(metric, ylabel, title):
    plt.figure(figsize=(13, 7))
    plotted = 0

    for window_name in prediction_sources:
        for model in selected_models:
            data = event_results_df[
                (event_results_df["Window"] == window_name) &
                (event_results_df["Model"] == model)
            ].sort_values("Lead Time (hours)")

            if data.empty:
                continue

            plt.plot(
                data["Lead Time (hours)"],
                data[metric],
                marker="o",
                linewidth=2,
                label=f"{model} | {window_name}"
            )
            plotted += 1

    if plotted == 0:
        plt.close()
        print("No lines to plot. Check event_results_df.")
        return

    plt.xlabel("Forecast lead time (hours)")
    plt.ylabel(ylabel)
    plt.title(title)
    plt.xticks(range(1, 13))
    plt.xlim(1, 12)
    plt.ylim(0, 1.05)
    plt.grid(True, alpha=0.3)
    plt.legend(bbox_to_anchor=(1.02, 1), loc="upper left")
    plt.tight_layout()
    plt.show()


plot_event_metric(
    "Recall",
    "Recall",
    "Low-Visibility Event Recall vs Forecast Lead Time"
)
```
